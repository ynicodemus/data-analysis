```python
import numpy as np
import pandas as pd 
import seaborn as sns 
import matplotlib.pyplot as plt 
import warnings 
warnings.filterwarnings('ignore')
```

```python
#datasets
terrorist = pd.read_csv('datasets\\Global_terrorist.csv', encoding='ISO-8859-1')
terrorist
```

```python
terrorist.columns
```

<h1>MIDDLE EAST & NORTH AFRICA BOMBING/EXPLOSION</h1>

```python
#filtering data into smaller scope

filter1 = terrorist[terrorist['region_txt'] == 'Middle East & North Africa']

filter2 = filter1[filter1['attacktype1_txt'] == 'Bombing/Explosion']

filter3 = filter2[filter2['nkill'] >= 10 ]

filter4 = filter3[filter3['iyear'] >= 2000]
```

<h1>DATA AGGREGATION</h1><i>means collecting and summarizing data from multiple sources</i>

```python
#aggregating data

ME_terror = filter4.groupby(['latitude', 'longitude']).size().reset_index(name = 'incident_count')
ME_terror
```

<h1>Visualization of Geospatial using FOLIUM</h1>

```python
import folium
from folium.plugins import HeatMap
```

```python
heat_df = ME_terror[['latitude', 'longitude', 'incident_count']].values.tolist()
```

```python
my_map = folium.Map(location = [45.9432, 24.9668], zoom_start = 4)
HeatMap(heat_df).add_to(my_map)
my_map
```

<h1>REGIONAL ANALYSIS</h1>

<h3>Which region has the highest number of facility/infrastructure attacks in asia?</h3>

<h1>South Asia</h1>

```python
south_a1 = terrorist[terrorist['region_txt'] == 'South Asia']

south_a2 = south_a1[south_a1['attacktype1_txt'] == 'Facility/Infrastructure Attack']
```

```python
#aggregating data

south_a = south_a2.groupby(['latitude', 'longitude']).size().reset_index(name = 'incident_count')
```

```python
south = south_a[['latitude', 'longitude', 'incident_count']].values.tolist()
print(south_a.count())
```

<h1>East Asia</h1>

```python
east_a1 = terrorist[terrorist['region_txt'] == 'East Asia']

east_a2 = east_a1[east_a1['attacktype1_txt'] == 'Facility/Infrastructure Attack']
```

```python
east_a = east_a2.groupby(['latitude', 'longitude']).size().reset_index(name = 'incident_count')
east_a = east_a.dropna(subset=['latitude', 'longitude'])
```

```python
east = east_a[['latitude', 'longitude', 'incident_count']].values.tolist()
print(east_a.count())
```

<h1>SouthEast Asia</h1>

```python
s_east_a1 = terrorist[terrorist['region_txt'] == 'Southeast Asia']

s_east_a2 = s_east_a1[s_east_a1['attacktype1_txt'] == 'Facility/Infrastructure Attack']
```

```python
s_east_a = s_east_a2.groupby(['latitude', 'longitude']).size().reset_index(name = 'incident_count')
s_east_a = s_east_a.dropna(subset=['latitude', 'longitude'])
```

```python
southeast = s_east_a[['latitude', 'longitude', 'incident_count']].values.tolist()
print(s_east_a.count())
```

<h1>Central Asia</h1>

```python
c_a1 = terrorist[terrorist['region_txt'] == 'Central Asia']

c_a2 = c_a1[c_a1['attacktype1_txt'] == 'Facility/Infrastructure Attack']
```

```python
c_a = c_a2.groupby(['latitude', 'longitude']).size().reset_index(name = 'incident_count')
c_a = c_a.dropna(subset=['latitude', 'longitude'])
```

```python
central = c_a[['latitude', 'longitude', 'incident_count']].values.tolist()
print(c_a.count())
```

```python
my_map = folium.Map(location = [20.0, 100.0], zoom_start = 3.5)

HeatMap(south).add_to(my_map)
HeatMap(east).add_to(my_map)
HeatMap(southeast).add_to(my_map)
HeatMap(central).add_to(my_map)
my_map
```
